```python
from pyspark.sql.functions import current_timestamp, lit
from delta.tables import DeltaTable
```

```python
# 1. DEFINE PATHS
# Fabric uses the 'Files' and 'Tables' abstraction
source_path = "Files/Landing/omexom_raw_assets.csv"
target_table = "bronze_assets"

print(f"Reading data from {source_path}...")
```

```python
# 2. READ RAW CSV
# We use header=True because our generator included column names
df_raw = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(source_path)
```

```python

# 3. ENRICH WITH METADATA
# Adding an 'IngestionTimestamp' is a best practice for auditing
df_bronze = df_raw.withColumn("IngestionTimestamp", current_timestamp()) \
                  .withColumn("SourceFile", lit("omexom_raw_assets.csv"))

# 4. SAVE AS DELTA TABLE (The Bronze Layer)
# 'overwrite' is fine for this demo, but in production, we often 'append'
df_bronze.write.format("delta").mode("overwrite").saveAsTable(target_table)

print(f"✅ Success! Data ingested into Bronze table: {target_table}")
```

```python
# 5. PREVIEW THE DATA
display(spark.table(target_table).limit(5))
```
